## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

I am using the content_refresh_anonymized.csv starter dataset to verify the volume of data available for clustering. The following code extracts 3 real numbers to show the count of high-impression pages and declining pages, proving there is enough variance to form meaningful clusters.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Loading the dataset from the repository
url = 'https://raw.githubusercontent.com/monis0908/Fly-Rank-Internship/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(url)

total_pages = len(df)
high_impressions = len(df[df['impressions_90d'] > 500])
declining_pages = len(df[df['trend_direction'] == 'down'])

print(f"Total content items in dataset: {total_pages}")
print(f"Pages with >500 impressions (90 days): {high_impressions}")
print(f"Pages with a declining trend: {declining_pages}")
```
